```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as mpatches

y = np.array([31, 29, 41])
mylabels = ["effusion",  "Tuberculosis", "Normal"]
mycolors = ['skyblue', 'lightgreen', 'gold']
legend_labels = ["1500", "1400", "2000"]  # Different legend labels
plt.pie(y, labels=mylabels, colors=mycolors, shadow=True)

# Adjusting the position of the current Axes object to center it
plt.gca().set_position([ 1,1,1, 1])


# Create legend patches with the same colors as the pie chart
legend_patches = [mpatches.Patch(color=color, label=label) for color, label in zip(mycolors, legend_labels)]

plt.legend(handles=legend_patches,title="Size(X_ray):", loc="center right", bbox_to_anchor=(1.4, 0.5))
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as mpatches

y = np.array([25, 25, 25, 25])
mylabels = ["effusion", "Mass", "Tuberculosis", "Normal"]
mycolors = ['skyblue', 'salmon', 'lightgreen', 'gold']
legend_labels = ["1800", "1800", "1800", "1800"]  # Different legend labels

# Plotting horizontal bars
plt.barh(mylabels, y, color=mycolors)

# Create legend patches with the same colors as the bar chart
legend_patches = [mpatches.Patch(color=color, label=label) for color, label in zip(mycolors, legend_labels)]

plt.legend(handles=legend_patches, title="Size(X_ray):", loc="center right", bbox_to_anchor=(1.4, 0.5))
plt.xlabel("Size")
plt.ylabel("Model")
plt.title("Distribution of X-ray Conditions")
plt.gca().invert_yaxis()  # Invert y-axis to have "Normal" at the top
plt.xticks([])
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

labels = ['svm', 'Random F', 'KNN']
sizes_1800 = [98, 95, 91]
sizes_1801 = [91, 90, 88]

# Stacked bar chart for 1800
bars_1800 = plt.barh(labels, sizes_1800, color='skyblue', label='LBB')

# Stacked bar chart for 1801, stacked on top of 1800
bars_1801 = plt.barh(labels, sizes_1801, color='salmon', left=sizes_1800, label='HOG')

plt.xlabel("Accuracy")
plt.ylabel("Model")
plt.title("CT")
plt.legend(title="Feature", loc="center right", bbox_to_anchor=(1.4, 0.5))
plt.gca().invert_yaxis()  # Invert y-axis to have "Normal" at the top

# Add number annotations for skyblue bars
for bar, size in zip(bars_1800, sizes_1800):
    plt.text(bar.get_width(), bar.get_y() + bar.get_height() / 2, f'{size}',
             va='center', ha='left', color='white', fontsize=8, fontweight='bold')

# Add number annotations for salmon bars
for bar, size in zip(bars_1801, sizes_1801):
    plt.text(bar.get_width(), bar.get_y() + bar.get_height() / 2, f'{size}',
             va='center', ha='right', color='white', fontsize=8, fontweight='bold')

plt.xticks([])

plt.show()
```

```python
import random
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Sample data generation
data = {
    "Crime Type Summer": [(["cnn", "svm", "knn", "random f", "decision tree"]) for _ in range(100)],
    "Crime Type Winter": [(["cnn", "svm", "knn", "random f", "decision tree"]) for _ in range(100)]
}
df = pd.DataFrame(data)

s = "Crime Type Summer|Crime Type Winter".split("|")

index = np.arange(5)
bar_width = 0.35

fig, ax = plt.subplots()

# Define colors
summer_color = 'orange'
winter_color = 'blue'

summer = ax.bar(index, df["Crime Type Summer"].value_counts(), bar_width,
                label="With Mass", color=summer_color)

winter = ax.bar(index+bar_width, df["Crime Type Winter"].value_counts(),
                 bar_width, label="Without Mass", color=winter_color)

ax.set_xlabel('Model')
ax.set_ylabel('Accuracy')
ax.set_title('With Mass & Without')
ax.set_xticks(index + bar_width / 2)
ax.set_xticklabels(["cnn", "svm", "knn", "random f", "decision tree"])
plt.legend(title="Feature", loc="center right", bbox_to_anchor=(1.4, 0.5))

# Numbers to display on top of each bar
summer_numbers = [96, 94, 74, 80, 43]
winter_numbers = [85, 84, 98, 66, 35]
plt.yticks([])

# Display numbers on top of each bar
for i, (summer_bar, winter_bar) in enumerate(zip(summer, winter)):
    ax.text(summer_bar.get_x() + summer_bar.get_width() / 2, summer_bar.get_height() + 1,
            summer_numbers[i], ha='center', va='bottom')
    ax.text(winter_bar.get_x() + winter_bar.get_width() / 2, winter_bar.get_height() + 1,
            winter_numbers[i], ha='center', va='bottom')

plt.show()
```
